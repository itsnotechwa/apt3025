```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
train = pd.read_csv('/content/titanic_train.csv')
```

```python
sns.heatmap(train.isnull(), yticklabels = False, cbar = True, cmap = 'viridis')
```

```python
train.head(10)
```

```python
# Compute mean ages for each Pclass
def compute_mean_ages(train):
    return train.groupby('Pclass')['Age'].mean()

# Function to impute missing Age values
def impute_age(cols, means):
    Age, Pclass = cols
    return Age if pd.notnull(Age) else means[Pclass]

# Compute mean ages for each class
mean_ages = compute_mean_ages(train)
```

```python
train['Age'] = train[['Age', 'Pclass']].apply(lambda x: round(impute_age(x, mean_ages), 1), axis =1)
```

```python
sns.heatmap(train.isnull(), yticklabels = False, cbar = True, cmap = 'viridis')
```

```python
train.head(10)
```

```python
sns.heatmap(train.isnull(), yticklabels = False, cbar = True, cmap = 'viridis')
```

```python
# Function to encode Cabin availability
def encode_cabin(cabin):
    return 0 if pd.isnull(cabin) else 1
```

```python
# Apply the function to the Cabin column
train['Cabin'] = train['Cabin'].apply(encode_cabin)
```

```python
train.head(10)
```

```python
sns.heatmap(train.isnull(), yticklabels = False, cbar = True, cmap = 'viridis')
```

```python
train.dropna(inplace = True)
```

```python
sns.heatmap(train.isnull(), yticklabels = False, cbar = True, cmap = 'viridis')
```

```python
sex = pd.get_dummies(train['Sex'], drop_first = True)
```

```python
embark = pd.get_dummies(train['Embarked'], drop_first = True)
```

```python
sex
```

```python
train.drop(['Sex', 'Embarked', 'Name', 'Ticket'], axis = 1, inplace = True)
```

```python
train = pd.concat([train, sex, embark], axis = 1)
```

BUILDING THE LOGISTIC REGRESSION MODEL


```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(train.drop('Survived', axis = 1), train['Survived'], test_size = 0.20, random_state = 101)
```
